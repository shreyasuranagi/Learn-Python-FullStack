# reverse a string by recursion

```python
def rev(s):
    if len(s)==1:
        return s 
    return rev(s[1:]) + s[0]

print(rev('hello'))
```

```python
#fibonacci - nth number

def fib(n):
    if n==0:
        return 0
    if n==1:
        return 1
    return fib(n-1) + fib(n-2)

fib(8)


#fibonacci - print n numbers using recursion

def f(n):
    if n==0:
        return 0
    if n==1:
        return 1
    
    return ( f(n-1) + f(n-2)) 

n=int(input('enter:'))
for i in range(n):
    print(f(i),end=' ')
```

```python
#reverse a number

n=int(input('enter:'))
rev=0
while n>0:
    d=n%10
    rev=rev*10+d 
    n=n//10 
print('reverse using while:',rev) 

# reverse using recursion
def reverse(n,rev=0):
    if n<=0:
        return rev
    
    d=n%10
    rev = rev*10 + d
    
    return reverse(n//10, rev)

print('reverse using recursion:')
reverse(142)
```

```python
#3 5 7 9 7 5 3

def mirror(i=3):
    if i==9:
        print(i,end= ' ')
        return
    print(i,end= ' ')
    mirror(i+2)
    print(i,end= ' ')
mirror()
```

```python
#calculate exponentiation 
def power(b,e):
    if e==1:
        return b
    return b*power(b,e-1)

print(power(3,4))
```

```python
#sum of digits of a number 

def sum(n,s=0):
    if n<=0:
        return s

    d=n%10
    s+=d 
    return sum(n//10,s)

sum(142)
```
