```python
#decorators

def instagram(func):
    def wrapper(*args,**kwargs):
        print('login')
        func(*args,**kwargs)
        print('logout')
    return wrapper 

@instagram
def shreya(name):
    print('shreya text with ',name)


shreya('nandhika')
```

```python
#decorators

def instagram(func):
    print('now in: instagram(func) wehre func is:',func)
    def wrapper(*args,**kwargs):
        print('now in: wrapper(*,**) where args[0] is:',args[0])
        print('login')
        print('now calling func in wrapper:',func)
        func(*args,**kwargs)
        print('logout')
    print('now returning wrapper after creation: ',wrapper)
    return wrapper 


print('before decorator---')
#below decorator @instagram executes first, then the actual function call executes
# below function shreya is replaced with shreya=instagram(shreya)(bcos of decorator)so actually, instagram gets called first
@instagram
def shreya(name):
    print('now in: shreya(name) where name is: ',name)
    print('shreya text with ',name)
print('after decorator---')

print("now calling ACTUAL function: shreya('nandhika') :",type(shreya))
shreya('nandhika')
```
