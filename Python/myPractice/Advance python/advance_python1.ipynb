4. Extract all prime numbers from a given collection of integers.

```python
c=[1,4,2,6,3]
def isprime(n):
    c=0
    if n==1:
        return False
    else:
        for i in range(1,int(n**0.5)+1):
            if n%i==0:
                c+=1
        return c==1
print(  list(filter( lambda n: isprime(n),  c )))
```

5. Find every unique vowel used across a collection of text sentences, ignoring case and spacing.

```python
sentences = ['I have t study', 'impleted Pythn', 'I practiced tday' ]

#low=[s.lower for s in sentences]
#print(list(low))

low= list(map(str.lower,sentences))
print(low)

vowels= { ch  for s in low for ch in s if ch in 'aeiou'}
print(vowels)
```
