4. Extract all prime numbers from a given collection of integers.

```python
c=[1,4,2,6,3]
def isprime(n):
    c=0
    if n==1:
        return False
    else:
        for i in range(1,int(n**0.5)+1):
            if n%i==0:
                c+=1
        return c==1
print(  list(filter( lambda n: isprime(n),  c )))
```
