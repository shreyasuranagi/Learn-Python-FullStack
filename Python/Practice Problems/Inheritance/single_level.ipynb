# Single level inheritance

```python
class HeadOffice:
    hname='penta'
    ceo='mr.ravi'
    hloc='btm'

    courses=['ds','ai']

    @classmethod
    def display(cls):
        print(cls.hname,cls.hloc,cls.ceo)

    def update_course(self,course):
        self.courses.append(course)

    @staticmethod
    def announce(*companies):
        print('Upcoming requirement schedulings:')
        for c in companies:
            print(c)

class BranchOffice(HeadOffice):
    bname = 'penta V'
    bloc='vijaynagar'

    def __init__(self,*courses):
        super().__init__()
        self.courses=courses
        

    def display_head(self):
        super().display()

    def display_branch(self):
        print(self.bname, self.bloc)

b=BranchOffice('PFS','JFS')
b2=BranchOffice('Testing','Automation')

print(b.bname)
print(b.ceo)
print(b.announce("abc","caltech","cognizent"))
print(HeadOffice.announce("abc","caltech","cognizent"))
print(b.courses)
print(b.update_course('DTA'))
```

class and object properties

```python
class HeadOffice:
    hname='penta'
    ceo='mr.ravi'
    hloc='btm'
    courses=['ds','ai']

    def display_head(self):
        print(self.hname, self.ceo, self.hloc, self.courses)

class BranchOffice(HeadOffice):

    def __init__(self,bname,bloc):
        self.bname=bname
        self.bloc=bloc

    def display_branch(self):
        print(self.bname, self.bloc)

b1=BranchOffice('penta v','vijaynagar')
b2=BranchOffice('penta h','hebbal')
#method to do display instead of printing each time

b1.display_branch()
b2.display_branch()
HeadOffice.display_head(HeadOffice())
b1.display_head()
b2.display_head()
HeadOffice.display_branch(HeadOffice())

```
