### SUMMARY

Lookup Hierarchy: Python checks the Instance Namespace (__dict__) first; if the attribute isn't there, it looks up in the Class Namespace.

Shadowing: Assigning a value to s1.college creates a new key in s1.__dict__ rather than overwriting the class attribute, protecting other instances.

Restoring Class Fallback: Using del s1.college deletes the instance-level override, allowing s1 to look up college from the class namespace once again.

Method Execution: s1.display() is syntactic sugar for passing the instance as the self argument into Student.display(s1).

````python
# ==========================================
# NAMESPACE & ATTRIBUTE DEMONSTRATION SCRIPT
# ==========================================

college_name = "ABC University"  # Global namespace

class Student:  # Global namespace
    college = "IIT"  # Class namespace

    def __init__(self, name, age):  # Class namespace
        self.name = name     # Instance namespace
        self.age = age       # Instance namespace

    def display(self):  # Class namespace
        message = "Student Details"  # Local namespace
        print(f"--- {message} ---")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"College: {self.college}")
        print()


# 1. INITIALIZING OBJECTS
s1 = Student("Rahul", 21)
s2 = Student("Priya", 22)

print("=== INITIAL STATE ===")
print("s1 namespace (__dict__):", s1.__dict__)
print("s2 namespace (__dict__):", s2.__dict__)
print("Student class namespace:", Student.college)
print("-" * 40)


# 2. ACCESSING ATTRIBUTES & HIERARCHY
print("=== 1. ACCESSING ATTRIBUTES ===")
# s1.college checks instance namespace first -> not found, checks class namespace -> finds "IIT"
print("s1.college:", s1.college) 
print("s2.college:", s2.college)

# Class namespace cannot access instance namespace (uncommenting below raises AttributeError)
# print(Student.name)  # Error: type object 'Student' has no attribute 'name'
print("(Note: Class namespace cannot access instance attributes like Student.name directly)\n")


# 3. MODIFYING ATTRIBUTES & SHADOWING
print("=== 2. MODIFYING & ATTRIBUTE SHADOWING ===")
# Modifying via instance: creates a new attribute in s1's namespace (shadowing class attribute)
s1.college = "MIT"
print("After s1.college = 'MIT':")
print("s1.__dict__ (Notice 'college' was created here):", s1.__dict__)
print("s1.college (resolves from instance):", s1.college)
print("s2.college (unaffected, still resolves from class):", s2.college)
print("Student.college (unaffected):", Student.college)
print("-" * 40)


# 4. MODIFYING VIA CLASS
print("=== 3. MODIFYING VIA CLASS ===")
# Changing at class level affects all objects that do NOT shadow it
Student.college = "NIT"
print("After Student.college = 'NIT':")
print("s1.college (still 'MIT' because it is shadowed in s1's namespace):", s1.college)
print("s2.college (changed to 'NIT' because it falls back to the class):", s2.college)
print("-" * 40)


# 5. REMOVING INSTANCE OVERRIDE (DELETING SHADOWED ATTR)
print("=== 4. REMOVING INSTANCE OVERRIDE (`del`) ===")
del s1.college
print("After 'del s1.college':")
print("s1.__dict__ (college key is removed):", s1.__dict__)
print("s1.college (now falls back to class namespace again):", s1.college)
print("-" * 40)


# 6. METHOD CALL EQUIVALENCE
print("=== 5. METHOD CALL RESOLUTION ===")
print("Calling s1.display():")
s1.display()

print("Calling Student.display(s1) explicitly (they do the exact same thing):")
Student.display(s1)
````

## mutable class atrr vs instance attr

````python
# ==========================================================
# MUTABLE CLASS ATTRIBUTES VS INSTANCE ATTRIBUTES IN PYTHON
# ==========================================================

print("=== SCENARIO A: Shared Class List ===")

class Student:
    # REASON 1: One list object referenced by the class attribute. 
    # All objects share this common data.
    subjects = []

    def __init__(self, name):
        self.name = name

s1 = Student("Rahul")
s2 = Student("Priya")

# 1. Appending to a shared list (In-place mutation)
# The first mutates the existing list object referenced by both.
s1.subjects.append("Math")
print("After s1.subjects.append('Math'):")
print("  s1.subjects:", s1.subjects)
print("  s2.subjects (sees change because it references the same list):", s2.subjects)
print("  Student.subjects:", Student.subjects)
print("-" * 50)


# 2. Replacing the list for one object (Shadowing via assignment)
# Assignment creates an instance attribute that shadows the class attribute.
s1.subjects = ["Physics", "Chemistry"]
print("After s1.subjects = ['Physics', 'Chemistry'] (Shadowing):")
print("  s1.__dict__ (Notice 'subjects' is now stored here):", s1.__dict__)
print("  s1.subjects (individual list):", s1.subjects)
print("  s2.subjects (unaffected, still points to class list):", s2.subjects)
print("  Student.subjects:", Student.subjects)
print("-" * 50)


# 3. Replacing the shared list at the class level
# You are changing the collection owned by the class.
Student.subjects = ["Computer Science", "English"]
print("After Student.subjects = ['Computer Science', 'English']:")
print("  s2.subjects (updated because it points to the class list):", s2.subjects)
print("  s1.subjects (unaffected because it is shadowed by its own instance list):", s1.subjects)
print("-" * 50)


print("=== SCENARIO B: Independent Instance Lists ===")

class IndependentStudent:
    def __init__(self, name):
        self.name = name
        # REASON 2: Each constructor call creates a separate list and instance binding.
        # Every student needs independent subject selections.
        self.subjects = []

ind1 = IndependentStudent("Amit")
ind2 = IndependentStudent("Neha")

ind1.subjects.append("Biology")
print("After ind1.subjects.append('Biology'):")
print("  ind1.subjects:", ind1.subjects)
print("  ind2.subjects (completely independent, unaffected):", ind2.subjects)
````

Lookup & Shadowing work the exact same way for everything (Python always checks the instance namespace first, then falls back to the class).

The Catch: Assignment (=) always creates/updates an instance attribute and shadows the class attribute. In-place mutation (.append(), .remove()) on a class attribute modifies the shared object directly unless that instance has already shadowed it with its own separate object.

## descriptors

````python
class Student:
    def __init__(self, age):
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Invalid age")
        self._age = value

s1 = Student(21)

# What if we forcefully bypass the setter and inject into __dict__?
s1.__dict__["age"] = -999 

print(s1.__dict__)  # Output: {'_age': 21, 'age': -999} (It's sitting right there in the instance dict!)

# Now, what happens when we access s1.age?
print(s1.age)  # Output: 21 (NOT -999!)
````

````python
# ==========================================================
# ADVANCED ATTRIBUTE LOOKUP & DESCRIPTOR DEMO SCRIPT
# ==========================================================

class Student:
    college = "IIT"  # Ordinary Class Attribute (Fallback Tier)

    def __init__(self, name, age):
        self.name = name      # Instance Attribute
        self._age = age       # Backing attribute for our property

    # 1. DATA DESCRIPTOR (@property + @setter)
    # Data descriptors occupy Lookup Priority #1 (Beat instance attributes!)
    @property
    def age(self):
        print("  --> [Data Descriptor] Property getter running for 'age'")
        return self._age

    @age.setter
    def age(self, value):
        print("  --> [Data Descriptor] Property setter running for 'age'")
        if value < 0:
            raise ValueError("Age cannot be negative!")
        self._age = value

    # 2. FALLBACK FOR MISSING ATTRIBUTES (__getattr__)
    # Invoked only after regular lookup fails (Lookup Priority #4)
    def __getattr__(self, attr_name):
        return f"  --> [__getattr__ fallback] Attribute '{attr_name}' does not exist. Handled dynamically!"


# ==========================================
# EXECUTION & TESTS
# ==========================================

if __name__ == "__main__":
    s1 = Student("Rahul", 21)

    print("=== TEST 1: DATA DESCRIPTOR PRECEDENCE ===" )
    print(f"s1.age access -> Value: {s1.age}")
    
    # Let's try to forcefully inject a fake age directly into s1.__dict__ 
    # to see if data descriptors really beat the instance dictionary:
    s1.__dict__["age"] = 999
    print("Injected 'age': 999 directly into s1.__dict__")
    print("s1.__dict__ contents:", s1.__dict__)
    
    # Even though it's sitting right in __dict__, Step 1 (Data Descriptor) intercepts it!
    print(f"Accessing s1.age now -> Value: {s1.age} (Notice it still ran the property getter!)")
    print("-" * 50)


    print("=== TEST 2: INSTANCE ATTRIBUTE VS CLASS ATTRIBUTE SHADOWING ===")
    print("Before shadowing:")
    print(f"  s1.college (falls back to class): {s1.college}")
    
    # Shadowing the class attribute by creating an instance attribute
    s1.college = "MIT"
    print("After s1.college = 'MIT' (Shadowing created):")
    print(f"  s1.college (from instance dict): {s1.college}")
    print(f"  Student.college (unaffected class attr): {Student.college}")
    print("-" * 50)


    print("=== TEST 3: REMOVING INSTANCE OVERRIDE (`del`) ===")
    del s1.college
    print("After 'del s1.college':")
    print(f"  s1.college (reveals inherited class attribute again): {s1.college}")
    print("-" * 50)


    print("=== TEST 4: NONEXISTENT ATTRIBUTE FALLBACK (__getattr__) ===")
    # Accessing an attribute that doesn't exist anywhere
    print(f"Accessing s1.gpa:")
    print(s1.gpa)
````

## CLASS STATIC INSTANCE METHODS

````python
# ==========================================================
# METHODS AND THEIR DIFFERENT TYPES DEMO SCRIPT
# ==========================================================

class Student:
    college = "IIT"  # Class attribute (shared by default)

    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age    # Instance attribute

    # 1. INSTANCE METHOD
    # Python automatically passes the instance (s1) as 'self'.
    # Used to work with a particular object's data.
    def display(self):
        print(f"  [Instance Method] Student Name: {self.name}, Age: {self.age}, College: {self.college}")

    # 2. CLASS METHOD
    # Python automatically passes the class (Student) as 'cls' (even if called via an instance).
    # Used to modify class-level data or serve as an alternative constructor.
    @classmethod
    def change_college(cls, new_college):
        print(f"  --> [Class Method] Changing class-level college to: {new_college}")
        cls.college = new_college

    # Alternative constructor example using a class method:
    @classmethod
    def from_string(cls, data_str):
        """Parses a comma-separated string like 'Rahul,21' to create a Student object."""
        name, age = data_str.split(",")
        return cls(name, int(age))

    # 3. STATIC METHOD
    # No automatic 'self' or 'cls' argument is supplied.
    # Used for utility operations related to the class concept, but needing no state.
    @staticmethod
    def is_adult(age):
        return age >= 18


# ==========================================================
# EXECUTION & TESTS
# ==========================================================

if __name__ == "__main__":
    print("=== 1. INITIALIZING OBJECTS ===")
    s1 = Student("Rahul", 21)
    s2 = Student("Priya", 17)

    s1.display()
    s2.display()
    print("-" * 50)

    print("=== 2. CALLING INSTANCE METHODS ===")
    # Method A: Calling through an instance (Python supplies s1 automatically)
    print("Calling s1.display():")
    s1.display()

    # Method B: Calling through the class (You must explicitly pass the instance as the argument)
    print("Calling Student.display(s1) explicitly:")
    Student.display(s1)
    print("-" * 50)

    print("=== 3. CLASS METHODS & CLASS-LEVEL MODIFICATION ===")
    # Changing the class attribute via the class method
    Student.change_college("MIT")

    print(f"s1.college (sees new class default): {s1.college}")
    print(f"s2.college (sees new class default): {s2.college}")
    print(f"Student.college: {Student.college}")
    print("-" * 50)

    print("=== 4. ALTERNATIVE CONSTRUCTOR (CLASS METHOD) ===")
    # Using the @classmethod as a factory/alternative constructor
    s3 = Student.from_string("Aman,22")
    print(f"Created via alternative constructor: {s3.name}, {s3.age}, College: {s3.college}")
    print("-" * 50)

    print("=== 5. STATIC METHODS ===")
    # Calling static methods via class or instance (no hidden state injected)
    print(f"Student.is_adult(21): {Student.is_adult(21)}")
    print(f"Student.is_adult(16): {Student.is_adult(16)}")
    print(f"s1.is_adult(25) (can also be called via instance): {s1.is_adult(25)}")
````

# INHERITANCE


````python
# ==========================================================
# INHERITANCE, MRO, AND POLYMORPHISM DEMO SCRIPT
# ==========================================================

class Parent:
    family_name = "Sharma"  # Parent class attribute

    def introduce(self):
        print(f"  [Parent method] Family Name: {self.family_name}")

    def show_branch_info(self):
        # POLYMORPHISM DEMO: 
        # A parent method can access a child-defined attribute through 'self' 
        # because 'self' points to the actual runtime instance (e.g., Child).
        branch_val = getattr(self, 'branch', "No branch defined")
        print(f"  [Parent method accessing self.branch] Branch: {branch_val}")


class Child(Parent):
    # Child class inherits from Parent. 
    # Notice: 'family_name' is NOT copied into Child.__dict__. It is resolved via MRO.

    def __init__(self, branch):
        self.branch = branch  # Child-specific instance attribute

    # 1. METHOD OVERRIDING
    def introduce(self):
        print("  --> [Child overridden introduce() START]")
        # 2. EXTENDING BEHAVIOR USING super()
        # super() searches after the current class in the MRO to call the parent version
        super().introduce()
        print(f"  --> [Child specific branch]: {self.branch}")
        print("  --> [Child overridden introduce() END]")


# ==========================================================
# EXECUTION & TESTS
# ==========================================================

if __name__ == "__main__":
    print("=== 1. INHERITANCE NAMESPACE & HIERARCHY INSPECTION ===")
    c = Child("Bangalore")
    
    print(f"Child.family_name (resolved via MRO): {Child.family_name}")
    print(f"Child.__dict__ (Notice family_name is NOT here): {Child.__dict__}")
    print(f"Child.__bases__ (Direct base classes): {Child.__bases__}")
    print(f"Child.__mro__ (Method Resolution Order): {Child.__mro__}")
    print(f"Parent.__subclasses__() (Live known subclasses): {Parent.__subclasses__()}")
    print("-" * 60)

    print("=== 2. METHOD OVERRIDING & super() ===")
    c.introduce()
    print("-" * 60)

    print("=== 3. THE POLYMORPHISM PRINCIPLE (`self` downward lookup) ===")
    # Calling a method defined strictly in Parent, which reads 'self.branch' defined in Child
    print("Calling c.show_branch_info():")
    c.show_branch_info()
    print("-" * 60)

    print("=== 4. MODIFICATION RULES (Parent vs Child) ===")
    print(f"Before modification - c.family_name: {c.family_name}")

    # Modifying at the Parent level affects descendants that don't shadow it
    Parent.family_name = "Sharma-Modified"
    print(f"After Parent.family_name = 'Sharma-Modified', c.family_name: {c.family_name}")

    # Modifying at the Child class level creates/rebinds an attribute in the child namespace
    Child.family_name = "Sharma-ChildSpecial"
    print(f"After Child.family_name = 'Sharma-ChildSpecial':")
    print(f"  Child.family_name: {Child.family_name}")
    print(f"  Parent.family_name (unaffected): {Parent.family_name}")
````

# single level inheritance

````python
# ==========================================================
# SINGLE INHERITANCE & CONSTRUCTOR COLLABORATION SCRIPT
# ==========================================================

class Animal:
    category = "Living Organism"  # Parent class attribute

    def __init__(self, name):
        self.name = name  # Parent-owned instance attribute

    def eat(self):
        print(f"  [Animal.eat] {self.name} is eating generic food.")


class Dog(Animal):
    breed_type = "Domestic"  # Child-specific class attribute

    def __init__(self, name, breed):
        # 1. Call the parent constructor to let the parent initialize its own state
        super().__init__(name)
        # 2. Child initializes its own specialized state
        self.breed = breed

    # 3. Overriding a parent method
    def eat(self):
        print(f"  --> [Dog.eat override START]")
        # 4. Executing parent implementation after overriding using super()
        super().eat()
        print(f"  --> [Dog.eat override END: {self.name} is crunching dog food!]")

    # 5. Child-specific method
    def bark(self):
        print(f"  [Dog.bark] {self.name} says Woof!")


# ==========================================
# EXECUTION & SCENARIO DEMONSTRATIONS
# ==========================================

if __name__ == "__main__":
    print("=== 1. CREATING INSTANCE & INSPECTING NAMESPACE (__dict__) ===")
    d1 = Dog("Tommy", "Labrador")
    
    # Notice how super().__init__(name) combined with child assignment populates d1.__dict__
    print(f"d1.__dict__: {d1.__dict__}")
    print(f"Dog inheritance base relationships (Dog.__bases__): {Dog.__bases__}")
    print("-" * 60)


    print("=== 2. ATTRIBUTE LOOKUP HIERARCHY (Parent vs Child) ===")
    # Access parent class attribute through child class and child object
    print(f"Dog.category (parent attr via child class): {Dog.category}")
    print(f"d1.category (parent attr via child object): {d1.category}")
    print(f"d1.name (parent instance attr): {d1.name}")
    print(f"d1.breed (child instance attr): {d1.breed}")
    print("-" * 60)


    print("=== 3. METHODS & OVERRIDING ===")
    # Inherited method and specialized child method
    d1.bark()  # Child-specific method
    print("Calling d1.eat() (Demonstrating method override + super().eat()):")
    d1.eat()  
    print("-" * 60)


    print("=== 4. WHAT HAPPENS IF YOU TRY TO CALL CHILD METHOD VIA PARENT? ===")
    # Animal has no concept of 'bark' because general animal category shouldn't assume it.
    try:
        Animal.bark(d1)
    except AttributeError as e:
        print(f"Caught expected error: {e}")
    print("-" * 60)


    print("=== 5. MODIFICATION RULES (Parent vs Child Class Defaults) ===")
    # Modify inherited class attribute through parent
    Animal.category = "Biological Entity"
    print(f"After Animal.category = 'Biological Entity' -> d1.category: {d1.category}")

    # Modify inherited class attribute through child (creates a child-level override)
    Dog.category = "Canine Pet"
    print(f"After Dog.category = 'Canine Pet':")
    print(f"  Dog.category: {Dog.category}")
    print(f"  Animal.category (unaffected): {Animal.category}")
    print("-" * 60)


    print("=== 6. SPECIAL CASE: THE COST OF OMITTING super().__init__() ===")
    
    class BadDog(Animal):
        def __init__(self, name, breed):
            # OMITTING super().__init__(name)!
            self.breed = breed

    bad_d1 = BadDog("Rocky", "Bulldog")
    print(f"bad_d1.__dict__: {bad_d1.__dict__}")
    try:
        print(bad_d1.name)  # Will raise AttributeError because Animal.__init__ never ran!
    except AttributeError as e:
        print(f"Caught expected error (Parent state missing): {e}")
````

# Multilevel inheritance

````python
# ==========================================================
# MULTILEVEL INHERITANCE & 3-TIER MRO DEMO SCRIPT
# ==========================================================

# Tier 1: Grandparent
class Animal:
    category = "Living Being"  # Grandparent class attribute

    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"  [Animal.eat] {self.name} (Grandparent level) is eating.")


# Tier 2: Parent (inherits from Animal)
class Dog(Animal):
    breed = "Domestic"  # Parent class attribute

    def __init__(self, name, breed):
        super().__init__(name)  # Relays to Animal.__init__
        self.breed = breed

    def bark(self):
        print(f"  [Dog.bark] {self.name} (Parent level) says Barking!")


# Tier 3: Child / Grandchild (inherits from Dog)
class Puppy(Dog):
    age = 1  # Child class attribute

    def __init__(self, name, breed, age):
        super().__init__(name, breed)  # Relays to Dog.__init__ (which relays to Animal)
        self.age = age

    def weep(self):
        print(f"  [Puppy.weep] {self.name} is weeping.")


# ==========================================================
# EXECUTION & TESTS
# ==========================================================

if __name__ == "__main__":
    p1 = Puppy("Max", "Golden Retriever", 0.5)

    print("=== 1. EXTENDED 3-TIER MRO INSPECTION ===")
    print(f"Puppy.__mro__: {Puppy.__mro__}")
    print(f"Puppy.__dict__ (Notice it only holds Puppy-specific keys): {p1.__dict__}")
    print("-" * 60)

    print("=== 2. TRANSITIVE MULTI-HOP LOOKUP ===")
    # p1 skips through Puppy -> Dog -> Animal to find 'category'
    print(f"p1.category (Grandparent attr): {p1.category}") 
    print(f"p1.breed (Parent attr): {p1.breed}")         
    print(f"p1.age (Child attr): {p1.age}")               
    print("-" * 60)

    print("=== 3. MULTI-TIER METHOD INHERITANCE ===")
    p1.eat()   # Found in Grandparent (Animal)
    p1.bark()  # Found in Parent (Dog)
    p1.weep()  # Found in Child (Puppy)
````

# multiple

````python
# ==========================================================
# ADVANCED MULTIPLE INHERITANCE & C3 LINEARIZATION SCRIPT
# ==========================================================

class Father:
    surname = "Sharma"

    def __init__(self, surname="Sharma", **kwargs):
        print("-> Father.__init__ running")
        self.father_surname = surname
        # Cooperative super(): passes remaining kwargs up the MRO chain
        super().__init__(**kwargs)

    def work(self):
        print("  [Father.work] Father works in IT.")

    def shared_skill(self):
        print("  [Father] Good at driving")


class Mother:
    language = "Hindi"

    def __init__(self, language="Hindi", **kwargs):
        print("-> Mother.__init__ running")
        self.mother_language = language
        super().__init__(**kwargs)

    def work(self):
        print("  [Mother.work] Mother works in Medicine.")

    def shared_skill(self):
        print("  [Mother] Good at cooking")


# Child inherits from Father first, then Mother
class Child(Father, Mother):
    age = 15

    def __init__(self, surname, language, age):
        print("-> Child.__init__ START")
        self.age = age
        # Cooperative call: triggers MRO sequence across BOTH parents
        super().__init__(surname=surname, language=language)
        print("-> Child.__init__ END")

    def play(self):
        print("  [Child.play] Child is playing video games.")


# ==========================================================
# EXECUTION & EDGE CASE DEMONSTRATIONS
# ==========================================================

if __name__ == "__main__":
    print("=== 1. INSPECTING THE MRO (Method Resolution Order) ===")
    # Notice the order: Child -> Father -> Mother -> object
    print("Child.__mro__:")
    for cls in Child.__mro__:
        print(f"  - {cls.__name__}")
    print("-" * 60)

    print("=== 2. COOPERATIVE INITIALIZATION (`super()` chain) ===")
    c1 = Child("Sharma", "Hindi", 15)
    print(f"c1 attributes initialized safely: Surname={c1.father_surname}, Language={c1.mother_language}, Age={c1.age}")
    print("-" * 60)

    print("=== 3. METHOD NAME COLLISION RESOLUTION (MRO Dominance) ===")
    print("Calling c1.work():")
    # Both Father and Mother have 'work()'. 
    # Because Father is first in class Child(Father, Mother), Father's version wins!
    c1.work()
    print("-" * 60)

    print("=== 4. ACCESSING THE SHADOWED PARENT METHOD EXPLICITLY ===")
    print("Calling Mother.work(c1) explicitly to bypass MRO priority:")
    Mother.work(c1)
````

# DIAMOND PROBLEM

````python

#### The Problem with Explicit Calls: The Diamond Problem


```python
class Human:
    def __init__(self):
        print("-> Human.__init__ running (Shared Base)")

class Father(Human):
    def __init__(self):
        print("-> Father.__init__ running")
        Human.__init__(self)  # Explicit call

class Mother(Human):
    def __init__(self):
        print("-> Mother.__init__ running")
        Human.__init__(self)  # Explicit call

class Child(Father, Mother):
    def __init__(self):
        print("-> Child.__init__ START")
        Father.__init__(self)   # Explicit call 1
        Mother.__init__(self)  # Explicit call 2
        print("-> Child.__init__ END")

c = Child()
````

```
class Child(Father, Mother):
    def __init__(self):
        super().__init__()  # Calls Father
        super().__init__()  # Calls Father AGAIN (restarts MRO lookup)
```

# solution to diamond probelm

````python
class Human:
    def __init__(self, **kwargs):
        print("-> Human.__init__ running")
        super().__init__(**kwargs)

class Father(Human):
    def __init__(self, surname, **kwargs):
        print("-> Father.__init__ running")
        self.surname = surname
        super().__init__(**kwargs)  # Seamlessly steps up to Mother/Human

class Mother(Human):
    def __init__(self, language, **kwargs):
        print("-> Mother.__init__ running")
        self.language = language
        super().__init__(**kwargs)

class Child(Father, Mother):
    def __init__(self, surname, language, age):
        print("-> Child.__init__ START")
        self.age = age
        super().__init__(surname=surname, language=language) # Triggers the entire chain cleanly once!
        print("-> Child.__init__ END")
````

# Hierarchical inheritance

````python
# ==========================================================
# HIERARCHICAL INHERITANCE DEMO SCRIPT
# ==========================================================

class Animal:
    category = "Living Being"  # Common parent attribute

    def eat(self):
        print("  [Animal.eat] Eating generic food.")


class Dog(Animal):
    breed = "Domestic"  # Dog-specific class attribute

    def bark(self):
        print("  [Dog.bark] Dog says: Woof!")

    # Child 1 can independently override the parent method
    def eat(self):
        print("  [Dog.eat] Dog is crunching on bones.")


class Cat(Animal):
    color = "White"  # Cat-specific class attribute

    def meow(self):
        print("  [Cat.meow] Cat says: Meow!")


# ==========================================================
# EXECUTION & TESTS
# ==========================================================

if __name__ == "__main__":
    d1 = Dog()
    c1 = Cat()

    print("=== 1. INHERITING COMMON PARENT ATTRIBUTES & METHODS ===")
    print(f"d1.category (Dog accessing parent): {d1.category}")
    print(f"c1.category (Cat accessing parent): {c1.category}")
    
    print("Calling inherited/overridden methods:")
    d1.eat()  # Uses Dog's overridden version
    c1.eat()  # Uses Animal's base version (since Cat didn't override it)
    print("-" * 60)


    print("=== 2. SPECIALIZED CHILD ATTRIBUTES & METHODS ===")
    print(f"d1.breed (Dog-specific): {d1.breed}")
    print(f"c1.color (Cat-specific): {c1.color}")

    d1.bark()
    c1.meow()
    print("-" * 60)


    print("=== 3. INDEPENDENCE TEST (Children don't mix!) ===")
    # Dog inherits from Animal, NOT from Cat. 
    # Attempting to call Cat methods on a Dog (or vice versa) throws an AttributeError.
    try:
        d1.meow()
    except AttributeError as e:
        print(f"Caught expected error (Dog has no 'meow'): {e}")

    try:
        c1.bark()
    except AttributeError as e:
        print(f"Caught expected error (Cat has no 'bark'): {e}")
    print("-" * 60)


    print("=== 4. CLASS-LEVEL MODIFICATION & SHADOWING ISOLATION ===")
    # Modifying the parent attribute through Dog creates a Dog-level override.
    Dog.category = "Canine Pet"
    
    print(f"Dog.category (modified): {Dog.category}")
    print(f"Cat.category (unaffected, still points to Animal): {Cat.category}")
    print(f"Animal.category (unaffected root): {Animal.category}")
    
    # c1 still sees the original parent value because Cat didn't shadow it
    print(f"c1.category: {c1.category}")
````

# HYBRID


````python
# ==========================================================
# HYBRID INHERITANCE & C3 LINEARIZATION MRO DEMO SCRIPT
# ==========================================================

# 1. Grandparent Class
class Animal:
    category = "Living Being"

    def eat(self):
        print("  [Animal.eat] Eating (Found via MRO search at Animal level).")


# 2. First Parent Class (inherits from Animal)
class Dog(Animal):
    breed = "Domestic"

    def bark(self):
        print("  [Dog.bark] Barking! (Found in Dog).")


# 3. Second Parent Class (inherits from Animal)
class Cat(Animal):
    color = "White"

    def meow(self):
        print("  [Cat.meow] Meowing! (Found in Cat).")


# 4. Specialized Child Class (Hybrid inherits from BOTH Dog and Cat)
class Hybrid(Dog, Cat):
    age = 1

    def play(self):
        print("  [Hybrid.play] Hybrid is playing specialized games.")

    # Overriding grandparent/parent method to show MRO / super() chain
    def eat(self):
        print("  --> [Hybrid.eat override START]")
        # super() looks up the next class in the MRO sequence (which is Dog, then Cat, then Animal)
        super().eat()
        print("  --> [Hybrid.eat override END]")


# ==========================================================
# EXECUTION & TESTS
# ==========================================================

if __name__ == "__main__":
    h1 = Hybrid()

    print("=== 1. INSPECTING THE HYBRID MRO CHAIN ===")
    print("Hybrid.__mro__:")
    for cls in Hybrid.__mro__:
        print(f"  - {cls.__name__}")
    print("-" * 60)


    print("=== 2. ATTRIBUTE LOOKUP ACROSS THE ENTIRE GRAPH ===")
    print(f"h1.category (Grandparent attr via Animal): {h1.category}")
    print(f"h1.breed (First Parent attr via Dog): {h1.breed}")
    print(f"h1.color (Second Parent attr via Cat): {h1.color}")
    print(f"h1.age (Child attr via Hybrid): {h1.age}")
    print("-" * 60)


    print("=== 3. METHOD RESOLUTION ORDER IN ACTION ===")
    print("Calling h1.eat() (Demonstrating MRO sequence traversal):")
    h1.eat()
    print()

    print("Calling first parent method (h1.bark()):")
    h1.bark()

    print("Calling second parent method (h1.meow()):")
    h1.meow()

    print("Calling child-specific method (h1.play()):")
    h1.play()
    print("-" * 60)


    print("=== 4. NAMESPACE CHECK (__dict__) ===")
    print(f"Hybrid.__dict__ (Only holds Hybrid-specific attributes/methods, not inherited ones):")
    print(list(Hybrid.__dict__.keys()))
````
