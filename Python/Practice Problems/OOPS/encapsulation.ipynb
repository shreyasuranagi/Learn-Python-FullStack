# using decorators

```python
class ScoreBoard:
    #inside class
    def __init__(self, name, challenges, score):
        self.name=name
        self._challenges=challenges
        self.__score=score

    @property
    def score(self):
        return self.__score

    @score.setter
    def score(self,points):
        self.__score += points

class Leetcode(ScoreBoard):
    def __init__(self,name,challenges,score,streak):
        super().__init__(name,challenges,score)
        self.streak=streak

    @streak.setter
    def streak(self,days):
        self.streak=days

#outside of the class
s1=Leetcode('playerabc',0,0,0)

print(s1.score)

s1.score=1050

print(s1.score)

#s1.streak=10
print(s1.streak)


```
